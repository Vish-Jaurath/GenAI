## 2️⃣ Definition of RAG

RAG (Retrieval-Augmented Generation) is a technique where relevant external information is first retrieved and then provided to the LLM before generating the final answer.


Instead of relying only on memory,
the LLM is allowed to:

- Search relevant information
- Read it
- Then answer



👉 RAG = Open-book exam for LLMs

Without RAG: model answers from memory

With RAG: model first looks at notes/documents

## 3️⃣ Why RAG is Needed

* RAG retrieves relevant information at runtime and injects it into the prompt as context
* In short: RAG = Retrieval + Prompting + Grounded Generation

❓ Q1. What is RAG and why is it needed?

👉 Answer:

RAG (Retrieval-Augmented Generation) is a framework where relevant external information is retrieved and added as context to the prompt before the LLM generates a response.

RAG is needed because LLMs have static knowledge and can hallucinate. Since LLMs generate probable tokens and not factual truth, they may confidently produce incorrect answers. RAG reduces hallucination by retrieving grounded information from external sources such as company PDFs or databases.

In RAG, internal company documents are converted into embeddings and stored externally in a vector database — the LLM itself does not store or memorize them.


------------

❓ Q2. Explain the complete RAG pipeline.

👉 Answer:

Query:
User asks a question, which is converted into embeddings.
Retrieve:
Document chunks are already stored as embeddings in a vector database. The query embedding is compared with stored embeddings using cosine similarity to retrieve the most semantically similar chunks.
Generate:
The retrieved text chunks are added as context to the prompt, and the LLM generates the final response token-by-token.


-----------

It's the process of splitting large documents into smaller meaningful pieces before creating embeddings.

## Practical

### ❓ Q1. What is a vector database and why is it needed in RAG systems?

RAG retrieves external information and provides it as context to the LLM to reduce hallucinations and improve factual accuracy.

For retrieval, documents are converted into embeddings, which are high-dimensional vectors. These embeddings need to be stored and searched efficiently, which is why vector databases are used.

Traditional databases are optimized for structured data, exact matching, and SQL queries, but they are not designed for semantic vector similarity search.

A vector database stores embeddings and performs efficient similarity retrieval using metrics such as cosine similarity.


### ❓ Q2. Explain how embeddings, cosine similarity, and vector DB work together during retrieval.

Initially, documents are chunked, and each chunk is converted into embeddings. These embeddings are stored in a vector database.

When a user asks a query, the query is also converted into an embedding.

The vector database compares the query embedding with stored embeddings using cosine similarity to identify semantically similar vectors.

Once the nearest vectors are retrieved, their corresponding original text chunks are fetched and added as context to the prompt. The LLM reads the retrieved context and generates the final response token-by-token using the attention mechanism.


### ❓ Q3. What is the difference between FAISS and a vector database?

FAISS is a library/framework developed by Meta for fast nearest-neighbor vector similarity search on large-scale embeddings.

It can store, index, and retrieve vectors efficiently, but it is not a complete database system.

A vector database, on the other hand, is a full-fledged system that provides vector storage, similarity search, APIs, metadata filtering, persistence, and scalability features.

In simple terms:

FAISS mainly focuses on fast vector search

Vector DB provides complete infrastructure around vector retrieval
