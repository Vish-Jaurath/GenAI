# 🚀 End-to-End RAG Project — Enterprise PDF Q&A Assistant

## Project Goal

Build a complete Retrieval-Augmented Generation (RAG) pipeline that:

- Loads PDF documents
- Chunks the text
- Creates embeddings
- Stores embeddings in FAISS vector DB
- Retrieves relevant chunks
- Sends context to LLM
- Generates grounded answers

## STEP 3 — UNDERSTAND WHAT HAPPENS INTERNALLY

1. Load documents

- Read text from files/PDFs.

2. Chunking

- Split large text into smaller pieces.

3. Embeddings

- Convert chunks into semantic vectors.

4. FAISS

- Store vectors and perform similarity search.

5. Retrieval

- Find chunks most similar to query.

6. LLM

- Use retrieved chunks as context and generate answer.

```python
Text files / Documents
→ chunking


Chunks
→ Gemini Embedding API
→ chunk embeddings


FAISS
→ stores/indexes embedding vectors

User query
→ Gemini Embedding API
→ query embedding


FAISS
→ cosine similarity search
→ retrieve relevant text chunks


Retrieved text chunks + query
→ final prompt


Gemini LLM
→ tokenization
→ token IDs
→ internal embeddings
→ attention over retrieved context
→ next-token prediction
→ response
```

External embedding API
Used for:
retrieval/search/similarity


Internal embeddings
Used for:
understanding + generation inside transformer

Low temperature (0–0.3): lower hallucination

High temperature (>0.7): hallucination risk increases

Lower temperature keeps RAG responses more grounded in retrieved context, while higher temperature increases creativity and hallucination risk.

#### VERY IMPORTANT

Overlap exists during: chunk creation phase

NOT retrieval phase directly.


Overlap is a chunking strategy, NOT a retrieval strategy

🧠 IMPORTANT

LLM does NOT see embeddings now.

It sees:

plain retrieved text

because transformers work on:

tokens/text

not vector DB embeddings directly.

RAG does not feed embeddings to LLM directly; it retrieves relevant text using embeddings and then gives that text as context to the LLM.
