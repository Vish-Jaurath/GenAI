## STEP 3 — UNDERSTAND WHAT HAPPENS INTERNALLY

1. Load documents

- Read text from files/PDFs.

2. Chunking

- Split large text into smaller pieces.

3. Embeddings

- Convert chunks into semantic vectors.

4. FAISS

- Store vectors and perform similarity search.

5. Retrieval

- Find chunks most similar to query.

6. LLM

- Use retrieved chunks as context and generate answer.

```python
Text files / Documents
→ chunking


Chunks
→ Gemini Embedding API
→ chunk embeddings


FAISS
→ stores/indexes embedding vectors

User query
→ Gemini Embedding API
→ query embedding


FAISS
→ cosine similarity search
→ retrieve relevant text chunks


Retrieved text chunks + query
→ final prompt


Gemini LLM
→ tokenization
→ token IDs
→ internal embeddings
→ attention over retrieved context
→ next-token prediction
→ response
```

External embedding API
Used for:
retrieval/search/similarity


Internal embeddings
Used for:
understanding + generation inside transformer

Low temperature (0–0.3): lower hallucination

High temperature (>0.7): hallucination risk increases

Lower temperature keeps RAG responses more grounded in retrieved context, while higher temperature increases creativity and hallucination risk.

#### VERY IMPORTANT

Overlap exists during: chunk creation phase

NOT retrieval phase directly.


Overlap is a chunking strategy, NOT a retrieval strategy

🧠 IMPORTANT

LLM does NOT see embeddings now.

It sees:

plain retrieved text

because transformers work on:

tokens/text

not vector DB embeddings directly.

RAG does not feed embeddings to LLM directly; it retrieves relevant text using embeddings and then gives that text as context to the LLM.

## LangChain 

Toolkit for building AI apps with LLMs and your own data.

LangChain is a Python framework that helps you build AI applications using Large Language Models (LLMs) like OpenAI GPT models.

In simple words: LangChain acts like a “bridge” between your AI model and your data/tools.


Without LangChain, you manually write lots of code to:
- load documents
- split text
- create embeddings
- search data
- connect to GPT
- maintain chat memory

LangChain gives ready-made components for all this.



Imagine building a chatbot for company policies.

- Without LangChain: You build every pipe, wire, and connection yourself.

- With LangChain: You get prebuilt LEGO blocks.

You simply connect them.

```python
# app.py

# Import libraries

import os
from dotenv import load_dotenv

# ==========================================
# LOAD API KEY
# ==========================================

load_dotenv()

GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")
print("GOOGLE_API_KEY LOADED")


# Document loading
from langchain_community.document_loaders import TextLoader

# Chunking
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Embeddings
from langchain_google_genai import GoogleGenerativeAIEmbeddings

# Vector DB
from langchain_community.vectorstores import FAISS

# Gemini LLM
from langchain_google_genai import ChatGoogleGenerativeAI

print("All dependencies have been imported...")


# ==========================================
# STEP 1 — LOAD DOCUMENTS
# ==========================================

print("Loading documents..")

files= [
    "data/hr_policy.txt",
    "data/insurance_policy.txt",
    "data/leave_policy.txt"
]

print(files)

all_docs= []

for file in files:
    loader= TextLoader(file)
    docs= loader.load()
    all_docs.extend(docs)

print(f"Loaded {len(all_docs)} documents")
# print(all_docs)

# ==========================================
# STEP 2 — CHUNKING
# ==========================================

print("Creating chunks...")

text_splitter= RecursiveCharacterTextSplitter(chunk_size= 300, chunk_overlap= 50)
chunks= text_splitter.split_documents(all_docs)
print(f"Total chunks created: {len(chunks)}")
print(chunks)

# ==========================================
# STEP 3 — CREATE EMBEDDINGS
# ==========================================

print("Creating embeddings...")

embedding_model = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001"
)

# # ==========================================
# # STEP 4 — STORE IN FAISS
# # ==========================================

print("Creating FAISS vector store...")

vector_store= FAISS.from_documents(chunks, embedding_model)

print("FAISS vector DB ready")
print(vector_store)


# ==========================================
# STEP 5 — LOAD GEMINI MODEL
# ==========================================

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.3
)

print("Gemini LLM loaded")
# print(llm)


# ========================================================
# STEP 6 — QUESTION ANSWERING LOOP/ QUERY + RETRIEVAL
# ========================================================

while True:
   
    query= input("\\n Ask Question: ")
    if query.lower == "exit":
        break

    print("\\n Searching relevant chunks...")

    # similarity retrieval

    docs= vector_store.similarity_search(query, k=2 # return top 3 most similar chunks
                                         )
    print(len(docs))
    # print(f"Question: {query}")
    # print(f"Similar chunks: {docs}")

    print("\\nRetrieved Chunks:\\n")

    for i, doc in enumerate(docs):
        print(f"Chunk {i+1}")
        print(doc.page_content)
        print("-" * 100)


    context= "\n\n".join([ doc.page_content for doc in docs])

    final_prompt= f"""
    Use the below context to answer the question.
    
    Context: {context}

    Question: {query}

    """
    
    print(f"This is the final prompt that passed to LLM: {final_prompt}")

    print(f"LLM starts working...")

    response= llm.invoke(final_prompt)

    # print(response)
    print("\n Final Answer:")
    print(response.content)
```
