## Transformers

A Transformer is a model that understands language by looking at all words at once and learning relationships between them.

A Transformer is a deep learning architecture designed to understand relationships between words in a sequence — no matter how far apart they are.

It’s the backbone of modern LLMs like
GPT
BERT

🧾 Example:

Sentence:
"The bank approved the loan because it had good credit history"

Transformer learns:
"it" → refers to customer
"bank" and "loan" are related
"credit history" influences "approved"

👉 It builds connections between words

💡 Old way (RNN/LSTM):
Read sentence word by word
Memory fades over long sentences

💡 Transformer way:
Reads entire sentence in parallel
Figures out which words matter to each other

Transformer =
👉 “Look at all words together + learn relationships”

## Attention Mechanism

This is the heart of Transformers

Attention = how much focus one word gives to another word

Each word asks: “Which other words should I pay attention to?”


👉 Attention = “relevance score between words”

### Technical intuition :

Each word creates:

Query → what I am looking for
Key → what I represent
Value → actual information

👉 Then it matches:
Query vs Key → decides importance

## Encoder vs Decoder

🧠 FINAL CORRECT UNDERSTANDING (Lock This)

👉 Tokens = word pieces
👉 Token IDs = numeric form of tokens
👉 Embeddings = dense vectors capturing meaning
👉 Encoder = understands context using attention
👉 Decoder = generates text using previous tokens

👉 Model does NOT understand:

text ❌
token IDs directly ❌

👉 Model understands: embeddings ✅

👉 Embeddings are NOT optional — they are the actual input to the Transformer


👉 Tokens → converted to numbers → converted to embeddings → THEN model understands


👉 Text → (optional preprocessing) → Tokenization → Tokens → Token IDs → Embeddings → Transformer (Encoder / Decoder)

Both encoder and decoder take token embeddings as input. 

Embeddings are created from token IDs before entering the Transformer, where attention mechanisms are applied for understanding (encoder) or generation (decoder).

| Step                  | Encoder                   | Decoder    |
| --------------------- | ------------------------- | ---------- |
| Input                 | embeddings                | embeddings |
| Focus                 | understanding             | generation |
| Uses previous output? | ❌ No                      | ✔ Yes      |
| Output                | contextual representation | next token |



Encoder does not use previous output because it processes the entire sentence at once and builds context in parallel, unlike the decoder which generates tokens sequentially using past outputs.

## BERT vs GPT difference

GPT is a true LLM used for text generation, while BERT is a transformer-based language model focused on understanding tasks and is not typically referred to as an LLM in practice.

All GPTs are LLMs, but not all Transformer models (like BERT) are considered LLMs in practice

⚙️ How they work
🧠 BERT (Bidirectional)

Looks left + right context

Understands full sentence at once

Example:

"bank" → uses full sentence to decide meaning

------------------------------------------------------

🤖 GPT (Unidirectional)

Looks left context only

Predicts next token

Example:

"I love drinking ___" → predicts “coffee”

👉 BERT
- Bidirectional → looks at left + right context together


🔥 Key Insight

- BERT uses full sentence at once
- So meaning is richer + complete

------------------------------------------

👉 GPT
- Unidirectional (causal) → looks only at left context
- GPT is predicting forward, not understanding full sentence

--------------------------------------------

GPT uses causal masking: Prevents seeing future tokens


BERT does NOT use masking like this: Sees full sentence

🧾 Example:

Sentence → "I love AI"

Focus word → "love"


| At word "love"       | BERT                 | GPT                    |
| -------------------- | -------------------- | ---------------------- |
| Can see left ("I")   | ✔                    | ✔                      |
| Can see right ("AI") | ✔                    | ❌                      |
| Goal                 | Understand meaning   | Predict next word      |
| Output               | Contextual embedding | Next token probability (Selecting a token from probability distribution) |

| Feature      | BERT          | GPT           |
| ------------ | ------------- | ------------- |
| Architecture\ Type | Encoder       | Decoder       |
| Context\ Direction     | Bidirectional | Left-to-right |
| Purpose\ Task     | Understanding | Generation    |
| Output       | Embeddings    | Tokens (text) |
| Example Tasks | Classification, Q&A | Chat, writing |

🧠 BERT (Encoder)

👉 Flow:
- Text → Tokens → IDs → Embeddings → Encoder (attention) → Output representation

- Looks at full sentence (left + right)
- Builds contextual embeddings
- Does NOT generate text


-------------------------------



🤖 GPT (Decoder)

👉 Flow:
- Text → Tokens → IDs → Embeddings → Decoder (attention + previous tokens) → Next token

- Looks left → right only
- Predicts next token
- Generates text sequentially

BERT is bidirectional and uses both left and right context to understand a word, whereas GPT is unidirectional and only uses past context to predict the next token.

GPT uses past context to predict a probability distribution over the next possible tokens and then generates one token by selecting from that distribution.


👉 BERT → understands (no generation)

👉 GPT → predicts probabilities → generates tokens

### LLMs don’t directly generate words — they generate probabilities, then pick words

❓ Q1. What is the key architectural difference between BERT and GPT?

👉 Answer:
BERT is an encoder-based Transformer used for understanding text, while GPT is a decoder-based Transformer used for generating text.
BERT produces contextual embeddings, whereas GPT predicts the next token probabilistically and generates sequences.

❓ Q2. Explain bidirectional vs unidirectional with an example.

👉 Answer:
Bidirectional (BERT) means the model uses both left and right context to understand a word.
Unidirectional (GPT) means the model uses only past (left) context to predict the next token.

Example: “I love AI” (focus = love)

BERT sees: “I” + “AI” → understands full meaning
GPT sees: “I” → predicts next token (AI, you, etc.)
❓ Q3. Why is BERT not suitable for text generation?

👉 Answer:
BERT is not suitable for generation because it processes the entire input bidirectionally and does not follow a sequential next-token prediction mechanism, which is required for generating text.

❓ Q4. What kind of tasks would you solve using BERT vs GPT?

👉 Answer:

BERT: classification, sentiment analysis, question answering, information extraction
GPT: text generation, chatbots, summarization, code generation
🚀 ROUND 2
❓ Q5. In GPT, why can’t the model see future tokens?

👉 Answer:
GPT uses causal masking, which prevents each token from attending to future tokens, ensuring proper left-to-right generation.

❓ Q6. What does GPT actually output at each step internally?

👉 Answer:
At each step, GPT outputs a probability distribution over all possible tokens, and then selects one token (via greedy or sampling) to generate the next word.

❓ Q7. What is the role of embeddings in both BERT and GPT?

👉 Answer:
Embeddings convert token IDs into dense numerical vectors that capture semantic meaning, and they serve as the actual input to the Transformer for both BERT and GPT.

Prompt engineering is the practice of designing input prompts to guide and control the output of LLMs effectively.

Zero-shot: no examples, only instructions

## Temperature
- Temperature controls how randomly the model selects the next token from the probability distribution



⚙️ Connect to what you know

GPT:
👉 predicts probability distribution over tokens

Example:
"I love"

| Token  | Probability |
| ------ | ----------- |
| AI     | 0.40        |
| you    | 0.30        |
| pizza  | 0.20        |
| coding | 0.10        |


🔹 Low Temperature (0 – 0.3)
- Makes distribution sharper

AI → very high chance
Others → almost ignored

✔ Output:

"I love AI" (almost always same)


---------------------------------

🔹 High Temperature (0.8 – 1)

👉 Makes distribution flatter

AI, you, pizza → more similar chances

✔ Output:

"I love pizza"
"I love coding"

👉 More creative / random


--------------------------
🧠 Intuition

- Low temp = safe, predictable
- High temp = creative, diverse



---------------------------------
❓ Is temperature always 0–1?

👉 ❌ No (important)

Common range: 0 to 1
But technically can go >1 (like 1.2, 1.5)

👉 Higher than 1 = very random / unstable output



🧠 Interview answer
- Temperature controls randomness in token selection by adjusting the probability distribution — lower values make outputs deterministic, higher values increase diversity.

## Max Tokens
- Maximum number of tokens the model can generate in output



⚙️ How it works
- Low max tokens → short answer
- High max tokens → long answer


🧠 Important

- Doesn’t affect quality directly
- Only limits length


🧾 Example

Prompt:

Explain AI

max_tokens = 20 → short answer
max_tokens = 200 → long answer





🔥 Perfect understanding

- Low temp → almost always same output
- High temp → can vary, but not guaranteed

🧠 One-line memory

- Temperature controls chance of variation, not certainty of variation


---------------------------------


📌 Context Window

👉 Total tokens the model can handle (input + output together)

🧾 Example

Context window = 100 tokens

Input = 80 tokens
Output can only be = 20 tokens


| Concept        | Meaning                         |
| -------------- | ------------------------------- |
| Max Tokens     | Output limit                    |
| Context Window | Total capacity (input + output) |



🧠 One-line clarity

- Max tokens = how much model can speak
- Context window = how much model can remember + speak together

--------------------------------

Final Combined Understanding

- Model predicts probabilities
- Temperature controls randomness of selection
- Max tokens controls output length
- Context window limits total memory
